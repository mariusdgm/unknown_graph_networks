# Alpha-bounds ablation — main-PC fixed comparator

Automatically uses the fresh `baseline_b_*` result directories when present, otherwise falls back to the historical August bounded runs on the main PC. Single-shot environment tables are reconstructed directly from the persistent cache if needed.


```python

from __future__ import annotations

import json
import math
from pathlib import Path
from typing import Any

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import t as student_t

pd.set_option("display.max_columns", 200)
pd.set_option("display.width", 260)


def find_repo_root(start: Path | None = None) -> Path:
    start = Path.cwd() if start is None else Path(start)
    for candidate in [start, *start.parents]:
        if (candidate / "opinion_dynamics").exists():
            return candidate
    raise RuntimeError("Could not find repository root containing opinion_dynamics/.")


REPO_ROOT = find_repo_root()
EXP_ROOT = REPO_ROOT / "opinion_dynamics" / "experiments"
RESULTS_ROOT = EXP_ROOT / "results"
OUT = RESULTS_ROOT / "alpha_bounds_comparison"
OUT.mkdir(parents=True, exist_ok=True)

# Prefer the fresh timing-baseline namespaces if they happen to exist.
# On the main PC the historical August result directories are the expected fallback.
SS_B_CANDIDATES = [
    RESULTS_ROOT / "baseline_b_ss",
    RESULTS_ROOT / "experiment_2026_08_17_broad_all_dynamics_multitopology_multiseed",
]
SS_U_CANDIDATES = [
    RESULTS_ROOT / "alpha_u_ss",
]
AB_B_CANDIDATES = [
    RESULTS_ROOT / "baseline_b_ab",
    RESULTS_ROOT / "experiment_2026_08_24_abundant_all_dynamics_multitopology_multiseed",
]
AB_U_CANDIDATES = [
    RESULTS_ROOT / "alpha_u_ab",
]


def choose_existing(candidates: list[Path], label: str) -> Path:
    for p in candidates:
        if p.exists():
            print(f"{label}: {p}")
            return p
    tried = "\n  ".join(str(p) for p in candidates)
    raise FileNotFoundError(f"Could not find {label}. Tried:\n  {tried}")


SS_B = choose_existing(SS_B_CANDIDATES, "single-shot bounded")
SS_U = choose_existing(SS_U_CANDIDATES, "single-shot unbounded")
AB_B = choose_existing(AB_B_CANDIDATES, "abundant bounded")
AB_U = choose_existing(AB_U_CANDIDATES, "abundant unbounded")


def mean_ci95(values):
    x = np.asarray(values, dtype=float)
    x = x[np.isfinite(x)]
    n = len(x)
    if n == 0:
        return np.nan, np.nan, np.nan
    mean = float(x.mean())
    if n == 1:
        return mean, mean, mean
    se = float(x.std(ddof=1) / math.sqrt(n))
    crit = float(student_t.ppf(0.975, df=n - 1))
    return mean, mean - crit * se, mean + crit * se


def _resolve_cache_dir(raw: str | Path, result_dir: Path, env_id: str, manifest: dict[str, Any]) -> Path:
    """Use manifest path first; if the repo moved, try local rebased cache locations."""
    raw_path = Path(raw)
    if raw_path.exists():
        return raw_path

    config_hash = str(manifest.get("config_hash", ""))
    study_name = result_dir.name
    candidates = [
        RESULTS_ROOT / "_trial_cache" / study_name / config_hash / env_id,
        RESULTS_ROOT / "_trial_cache" / study_name / env_id,
    ]
    for p in candidates:
        if p.exists():
            return p

    # Last-resort search under this study's trial-cache root.
    root = RESULTS_ROOT / "_trial_cache" / study_name
    if root.exists():
        matches = list(root.glob(f"**/{env_id}"))
        if len(matches) == 1:
            return matches[0]

    return raw_path


def load_single_shot_env(result_dir: Path, label: str) -> pd.DataFrame:
    csv_path = result_dir / "environment_level_results.csv"
    if csv_path.exists():
        df = pd.read_csv(csv_path)
        print(f"{label}: loaded {len(df)} environment rows from {csv_path.name}")
        return df

    manifest_path = result_dir / "RUN_MANIFEST.json"
    if not manifest_path.exists():
        raise FileNotFoundError(
            f"{label}: neither environment_level_results.csv nor RUN_MANIFEST.json exists in {result_dir}"
        )

    print(f"{label}: environment_level_results.csv absent; reconstructing directly from cache.")
    manifest = json.loads(manifest_path.read_text(encoding="utf-8"))
    seeds = [int(s) for s in manifest["learning_seeds"]]

    baseline_rows = []
    learned_rows = []
    missing = []

    for env_meta in manifest["environments"]:
        env_id = env_meta["environment_id"]
        cache_dir = _resolve_cache_dir(env_meta["cache_dir"], result_dir, env_id, manifest)

        baseline_file = cache_dir / "baseline_summary.json"
        if not baseline_file.exists():
            missing.append((env_id, "baseline", str(cache_dir)))
            continue
        baseline_rows.extend(json.loads(baseline_file.read_text(encoding="utf-8")))

        learned_root = cache_dir / "learned"
        for seed in seeds:
            candidates = (
                list(learned_root.glob(f"seed_{seed:02d}_*/summary.json"))
                if learned_root.exists()
                else []
            )
            if len(candidates) != 1:
                missing.append((env_id, f"learned_seed_{seed}", str(cache_dir)))
                continue
            learned_rows.append(json.loads(candidates[0].read_text(encoding="utf-8")))

    if missing:
        print("First missing entries:", missing[:10])
        raise RuntimeError(f"{label}: cache reconstruction incomplete ({len(missing)} missing entries).")

    baseline_df = pd.DataFrame(baseline_rows)
    learned_df = pd.DataFrame(learned_rows)

    learned_env = (
        learned_df.groupby(
            ["environment_id", "scenario_class", "family", "dynamics"],
            as_index=False,
        )
        .agg(
            learned_final_mean=("final_mean", "mean"),
            learned_final_mean_seed_std=("final_mean", "std"),
            final_fit_model_over_identity=("final_fit_model_over_identity", "mean"),
        )
    )

    def baseline_policy(policy: str, prefix: str) -> pd.DataFrame:
        return (
            baseline_df[baseline_df["policy"] == policy][["environment_id", "final_mean"]]
            .rename(columns={"final_mean": f"{prefix}_final_mean"})
        )

    env = (
        learned_env
        .merge(baseline_policy("uniform", "uniform"), on="environment_id", validate="one_to_one")
        .merge(baseline_policy("true_graph", "true_graph"), on="environment_id", validate="one_to_one")
        .merge(baseline_policy("no_control", "no_control"), on="environment_id", validate="one_to_one")
    )
    print(f"{label}: reconstructed {len(env)} environment rows.")
    return env


def load_abundant_env(result_dir: Path, label: str) -> pd.DataFrame:
    combined_path = result_dir / "combined_summary.csv"
    if not combined_path.exists():
        raise FileNotFoundError(
            f"{label}: missing {combined_path}. Run the experiment notebook once in non-worker/check mode to merge the completed cache."
        )

    df = pd.read_csv(combined_path)
    required = {"environment_id", "scenario_class", "family", "dynamics", "policy", "mean_end"}
    missing_cols = sorted(required - set(df.columns))
    if missing_cols:
        raise RuntimeError(f"{label}: combined_summary.csv missing columns: {missing_cols}")

    learned = df[df["policy"].astype(str).str.contains("learned", case=False, na=False)].copy()
    uniform = df[df["policy"].astype(str).str.fullmatch("uniform", case=False, na=False)].copy()
    true_graph = df[df["policy"].astype(str).str.fullmatch("true_graph", case=False, na=False)].copy()
    no_control = df[df["policy"].astype(str).str.fullmatch("no_control", case=False, na=False)].copy()

    agg_spec = {
        "learned_final_mean": ("mean_end", "mean"),
        "learned_final_mean_seed_std": ("mean_end", "std"),
    }
    if "model_over_identity" in learned.columns:
        agg_spec["prediction_ratio"] = ("model_over_identity", "mean")

    learned_env = (
        learned.groupby(["environment_id", "scenario_class", "family", "dynamics"], as_index=False)
        .agg(**agg_spec)
    )

    def base_env(frame: pd.DataFrame, name: str) -> pd.DataFrame:
        return (
            frame.groupby("environment_id", as_index=False)["mean_end"]
            .mean()
            .rename(columns={"mean_end": f"{name}_final_mean"})
        )

    env = (
        learned_env
        .merge(base_env(uniform, "uniform"), on="environment_id", validate="one_to_one")
        .merge(base_env(true_graph, "true_graph"), on="environment_id", validate="one_to_one")
        .merge(base_env(no_control, "no_control"), on="environment_id", validate="one_to_one")
    )
    print(f"{label}: loaded {len(env)} environment rows from combined_summary.csv")
    return env


ss_b = load_single_shot_env(SS_B, "bounded single-shot")
ss_u = load_single_shot_env(SS_U, "unbounded single-shot")
ab_b = load_abundant_env(AB_B, "bounded abundant")
ab_u = load_abundant_env(AB_U, "unbounded abundant")


def compare_regime(bounded: pd.DataFrame, unbounded: pd.DataFrame, regime: str):
    keys = ["environment_id", "scenario_class", "family", "dynamics"]

    bcols = keys + [
        "learned_final_mean",
        "uniform_final_mean",
        "true_graph_final_mean",
        "no_control_final_mean",
    ]
    ucols = bcols.copy()

    pred_b = "final_fit_model_over_identity" if "final_fit_model_over_identity" in bounded.columns else (
        "prediction_ratio" if "prediction_ratio" in bounded.columns else None
    )
    pred_u = "final_fit_model_over_identity" if "final_fit_model_over_identity" in unbounded.columns else (
        "prediction_ratio" if "prediction_ratio" in unbounded.columns else None
    )
    if pred_b:
        bcols.append(pred_b)
    if pred_u:
        ucols.append(pred_u)

    b = bounded[bcols].copy()
    u = unbounded[ucols].copy()

    ren_b = {
        "learned_final_mean": "bounded_learned",
        "uniform_final_mean": "bounded_uniform",
        "true_graph_final_mean": "bounded_true_graph",
        "no_control_final_mean": "bounded_no_control",
    }
    ren_u = {
        "learned_final_mean": "unbounded_learned",
        "uniform_final_mean": "unbounded_uniform",
        "true_graph_final_mean": "unbounded_true_graph",
        "no_control_final_mean": "unbounded_no_control",
    }
    if pred_b:
        ren_b[pred_b] = "bounded_prediction_ratio"
    if pred_u:
        ren_u[pred_u] = "unbounded_prediction_ratio"

    b = b.rename(columns=ren_b)
    u = u.rename(columns=ren_u)

    pair = b.merge(u, on=keys, how="inner", validate="one_to_one")
    if len(pair) != len(bounded) or len(pair) != len(unbounded):
        raise RuntimeError(
            f"{regime}: pairing mismatch: bounded={len(bounded)}, unbounded={len(unbounded)}, paired={len(pair)}"
        )

    # The environment/baseline side of the experiment should be identical.
    baseline_checks = {}
    for name in ["uniform", "true_graph", "no_control"]:
        delta = np.abs(pair[f"unbounded_{name}"] - pair[f"bounded_{name}"])
        baseline_checks[name] = float(delta.max())
    print(f"\n{regime} baseline max absolute differences:", baseline_checks)
    if max(baseline_checks.values()) > 1e-10:
        print("WARNING: baseline results are not numerically identical across bounded/unbounded runs.")

    pair["delta_unbounded_minus_bounded"] = pair["unbounded_learned"] - pair["bounded_learned"]
    pair["bounded_gain_vs_uniform"] = pair["bounded_learned"] - pair["bounded_uniform"]
    pair["unbounded_gain_vs_uniform"] = pair["unbounded_learned"] - pair["unbounded_uniform"]
    pair["delta_gain_vs_uniform"] = pair["unbounded_gain_vs_uniform"] - pair["bounded_gain_vs_uniform"]

    if "bounded_prediction_ratio" in pair.columns and "unbounded_prediction_ratio" in pair.columns:
        pair["delta_prediction_ratio"] = (
            pair["unbounded_prediction_ratio"] - pair["bounded_prediction_ratio"]
        )

    rows = []
    for (scenario, dynamics), g in pair.groupby(["scenario_class", "dynamics"], sort=False):
        dm, dlo, dhi = mean_ci95(g["delta_unbounded_minus_bounded"])
        gm, glo, ghi = mean_ci95(g["delta_gain_vs_uniform"])
        row = {
            "regime": regime,
            "scenario_class": scenario,
            "dynamics": dynamics,
            "n_environments": len(g),
            "bounded_learned_mean": float(g["bounded_learned"].mean()),
            "unbounded_learned_mean": float(g["unbounded_learned"].mean()),
            "delta_learned_mean": dm,
            "delta_learned_ci_low": dlo,
            "delta_learned_ci_high": dhi,
            "bounded_gain_vs_uniform_mean": float(g["bounded_gain_vs_uniform"].mean()),
            "unbounded_gain_vs_uniform_mean": float(g["unbounded_gain_vs_uniform"].mean()),
            "delta_gain_vs_uniform_mean": gm,
            "delta_gain_vs_uniform_ci_low": glo,
            "delta_gain_vs_uniform_ci_high": ghi,
            "environment_win_rate_unbounded_vs_bounded": float(
                (g["delta_unbounded_minus_bounded"] > 0).mean()
            ),
        }
        if "delta_prediction_ratio" in g.columns:
            pm, plo, phi = mean_ci95(g["delta_prediction_ratio"])
            row.update({
                "bounded_prediction_ratio_mean": float(g["bounded_prediction_ratio"].mean()),
                "unbounded_prediction_ratio_mean": float(g["unbounded_prediction_ratio"].mean()),
                "delta_prediction_ratio_mean": pm,
                "delta_prediction_ratio_ci_low": plo,
                "delta_prediction_ratio_ci_high": phi,
            })
        rows.append(row)

    summary = pd.DataFrame(rows)

    # Overall paired result across all 195 environments.
    om, olo, ohi = mean_ci95(pair["delta_unbounded_minus_bounded"])
    overall = pd.DataFrame([{
        "regime": regime,
        "scenario_class": "ALL",
        "dynamics": "ALL",
        "n_environments": len(pair),
        "bounded_learned_mean": float(pair["bounded_learned"].mean()),
        "unbounded_learned_mean": float(pair["unbounded_learned"].mean()),
        "delta_learned_mean": om,
        "delta_learned_ci_low": olo,
        "delta_learned_ci_high": ohi,
        "environment_win_rate_unbounded_vs_bounded": float(
            (pair["delta_unbounded_minus_bounded"] > 0).mean()
        ),
    }])

    # Structured-family diagnostic.
    fam_rows = []
    structured = pair[pair["scenario_class"] == "structured_mechanism"]
    for (dynamics, family), g in structured.groupby(["dynamics", "family"], sort=False):
        dm, dlo, dhi = mean_ci95(g["delta_unbounded_minus_bounded"])
        fam_rows.append({
            "regime": regime,
            "dynamics": dynamics,
            "family": family,
            "n_environments": len(g),
            "bounded_learned_mean": float(g["bounded_learned"].mean()),
            "unbounded_learned_mean": float(g["unbounded_learned"].mean()),
            "delta_learned_mean": dm,
            "delta_ci_low": dlo,
            "delta_ci_high": dhi,
        })
    family_summary = pd.DataFrame(fam_rows)

    return pair, summary, overall, family_summary


ss_pair, ss_summary, ss_overall, ss_family = compare_regime(ss_b, ss_u, "single_shot")
ab_pair, ab_summary, ab_overall, ab_family = compare_regime(ab_b, ab_u, "abundant")

ss_pair.to_csv(OUT / "single_shot_environment_level_comparison.csv", index=False)
ab_pair.to_csv(OUT / "abundant_environment_level_comparison.csv", index=False)
ss_summary.to_csv(OUT / "single_shot_bounded_vs_unbounded_by_dynamics.csv", index=False)
ab_summary.to_csv(OUT / "abundant_bounded_vs_unbounded_by_dynamics.csv", index=False)
ss_family.to_csv(OUT / "single_shot_structured_family_comparison.csv", index=False)
ab_family.to_csv(OUT / "abundant_structured_family_comparison.csv", index=False)

combined = pd.concat([ss_overall, ab_overall, ss_summary, ab_summary], ignore_index=True)
combined.to_csv(OUT / "alpha_bounds_ablation_compact.csv", index=False)

print("\n=== OVERALL ===")
display(pd.concat([ss_overall, ab_overall], ignore_index=True).round(6))

print("\n=== BY DYNAMICS / SCENARIO ===")
display(pd.concat([ss_summary, ab_summary], ignore_index=True).round(6))

print("\n=== STRUCTURED FAMILIES ===")
display(pd.concat([ss_family, ab_family], ignore_index=True).round(6))

plot_df = pd.concat([ss_summary, ab_summary], ignore_index=True).copy()
plot_df["label"] = (
    plot_df["regime"] + " | " +
    plot_df["scenario_class"] + " | " +
    plot_df["dynamics"]
)
fig, ax = plt.subplots(figsize=(9, max(5, 0.42 * len(plot_df))))
y = np.arange(len(plot_df))
ax.barh(y, plot_df["delta_learned_mean"].to_numpy(float))
ax.axvline(0.0, linewidth=1)
ax.set_yticks(y)
ax.set_yticklabels(plot_df["label"])
ax.invert_yaxis()
ax.set_xlabel("Final mean opinion: unbounded - bounded")
ax.set_title("Alpha-bound ablation")
plt.tight_layout()
plt.show()

print("\nComparison complete.")
print("Outputs:", OUT)
print("Main decision table:", OUT / "alpha_bounds_ablation_compact.csv")
```
